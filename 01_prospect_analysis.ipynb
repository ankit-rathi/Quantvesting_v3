## Prospect Analysis

Interactive prospect-selection view backed entirely by the Quantvesting engine.

```python
# 1. Environment & imports

from google.colab import drive
drive.mount('/content/drive')

import sys

PROJECT = "/content/drive/My Drive/quantvesting_v3"
sys.path.insert(0, PROJECT + "/src")

from quantvesting import (
    Quantvesting,
    load_config,
    load_all_data,
    PROSPECT_DISPLAY_COLUMNS,
)
```

```python
# 2. Load configuration and data

config = load_config(
    PROJECT + "/config/strategy.yaml"
)

data = load_all_data(
    PROJECT + "/data"
)

qv = Quantvesting(config)
```

```python
# 3. Run prospect analysis
# Keep portfolio membership visible so you can distinguish new prospects
# from stocks already held in DM/SV.

df_prospects = qv.prospects(
    data,
    include_portfolio=True,
)
```

```python
# 4. Interactive prospect view

qv.display_dataframe(
    df_prospects,
    columns=PROSPECT_DISPLAY_COLUMNS,
    sort_by="CumlRnk",
    ascending=True,
)
```

### Optional decision layer

```python
# 5. Candidate actions

df_prospects_actions = qv.prospect_actions(
    df_prospects,
    top_n=10,
)

qv.display_dataframe(
    df_prospects_actions,
    columns=PROSPECT_DISPLAY_COLUMNS,
    sort_by="CumlRnk",
    ascending=True,
)
```
