```python
# function to get booked and reserved amount
def get_amt():
  # fetch the JSON data from the URL
  url = "https://raw.githubusercontent.com/ankit-rathi/Tradevesting_v1/main/data/myPortfolioAmts.json"
  response = requests.get(url)
  pf_amts = response.json()  # parse the JSON data

  # extract values from the JSON
  py_booked_amt_dm = pf_amts["py_booked_amt_dm"]
  py_booked_amt_sv = pf_amts["py_booked_amt_sv"]
  cy_booked_amt_dm = pf_amts["cy_booked_amt_dm"]
  cy_booked_amt_sv = pf_amts["cy_booked_amt_sv"]
  reserve_amt_dm = pf_amts["reserve_amt_dm"]
  reserve_amt_sv = pf_amts["reserve_amt_sv"]

  # perform calculations
  py_booked_amt = py_booked_amt_dm + py_booked_amt_sv
  cy_booked_amt = cy_booked_amt_dm + cy_booked_amt_sv
  reserve_amt = reserve_amt_dm + reserve_amt_sv
  total_booked_amt = py_booked_amt + cy_booked_amt

  return total_booked_amt, reserve_amt, cy_booked_amt, py_booked_amt

gen_output = 0
```

```python
# function to format the amount
def format_amt(number):
    abs_number = abs(number)

    if abs_number >= 1_00_00_000:  # Crores
        formatted_amt = f"{number / 1_00_00_000:.2f} C"
    elif abs_number >= 1_00_000:  # Lakhs
        formatted_amt = f"{number / 1_00_000:.2f} L"
    elif abs_number >= 1_000:  # Thousands
        formatted_amt = f"{number / 1_000:.2f} K"
    else:
        formatted_amt = f"{number:.2f}"

    return formatted_amt

# function to set start and end date
def get_start_end_date():
  start_date = (datetime.date.today() + datetime.timedelta(days=-365)).strftime('%Y-%m-%d')
  end_date = (datetime.date.today() + datetime.timedelta(days=1)).strftime('%Y-%m-%d')
  return start_date, end_date

# function to fetch my portfolio csv
def get_mypfs_df():
  mypfs_df = pd.read_csv('https://raw.githubusercontent.com/ankit-rathi/Tradevesting_v1/main/data/myPortfolioStocks.csv')
  return mypfs_df

# function to fetch my prospects csv
def get_mypps_df():
  mypps_df = pd.read_csv('https://raw.githubusercontent.com/ankit-rathi/Tradevesting_v1/main/data/myProspectsScrips.csv')
  return mypps_df

# function to fetch screener data
def get_myscreen_df():
  myscreen_df = pd.read_csv('https://raw.githubusercontent.com/ankit-rathi/Tradevesting_v1/main/data/myScreenerDB.csv')
  return myscreen_df

# function to fetch momentum data
def get_myinvmt_df():
  myinvmt_df = pd.read_csv('https://raw.githubusercontent.com/ankit-rathi/Tradevesting_v1/main/data/myInvestments.csv')
  return myinvmt_df

# function to get the stock ids
def get_stock_ids(df_pf):
  stock_n100 = df_pf['Symbol'].unique()

  exclude = ['CADILAHC','MMTC', 'MASFIN']

  stock_ids = df_pf[~df_pf['Symbol'].isin(exclude) ]['Symbol'].unique()

  #mypf = mypf[mypf.Forecast.notnull()]
  #stock_ids = mypf['Symbol'].unique()

  #stock_ids.sort()
  return stock_ids

# get features from screener data
def get_screener_features():
  myscreen_df = get_myscreen_df()
  cols = ['Symbol', 'EPS', 'MedPE', 'ROCE%', 'ROE%', 'CapType']
  return myscreen_df[cols]

# get relative strength
def get_relative_strength(stock_list):

    # Dictionary to store stock tickers and their corresponding percentage price change
    stock_changes = {}

    # Loop through each stock and fetch its price data
    for stock in stock_list:
        try:
            # Download the stock data for the given date range
            stock_data = yfin.Ticker(stock + '.NS').history(period='1mo', interval='1d')[map(str.title, ['open', 'close', 'low', 'high', 'volume'])]

            # Calculate the percentage change for the stock
            if len(stock_data) > 0:
                start_price = stock_data['Close'].iloc[0]
                end_price = stock_data['Close'].iloc[-1]
                percent_change = round(((end_price - start_price) / start_price) * 100, 2)
                stock_changes[stock] = percent_change
            else:
                stock_changes[stock] = np.nan  # If no data is available, set to NaN

        except Exception as e:
            print(f"Error fetching data for {stock}: {e}")
            stock_changes[stock] = np.nan

    # Create a DataFrame with stock tickers and their percentage changes
    df = pd.DataFrame(list(stock_changes.items()), columns=['Symbol', 'Percent_Change'])

    # Drop any stocks with missing data (NaN values)
    df = df.dropna()

    # Calculate the percentile rank based on percentage change
    df['RSP'] = round(df['Percent_Change'].rank(pct=True) * 100, 2)
    cols = ['Symbol', 'RSP']
    df = df[cols]
    # Sort by percentile rank (optional)
    df = df.sort_values(by='RSP', ascending=False).reset_index(drop=True)

    return df
```

```python
# function to get stock technicals
def stock_prec_dev(stock_symbol):
    #stock_symbol = 'ULTRACEMCO.NS'
    short_window = 20
    mid_window = 50
    long_window = 200
    moving_avg = 'SMA'

    start = datetime.datetime(*map(int, start_date.split('-')))
    end = datetime.datetime(*map(int, end_date.split('-')))

    stock_df = yfin.Ticker(stock_symbol + '.NS').history(period='max', interval='1d')[map(str.title, ['open', 'close', 'low', 'high', 'volume'])]

    stock_df['Max'] = round(max(stock_df['Close']),2)
    stock_df = stock_df[(stock_df.index <= end_date) & (stock_df.index >= start_date)]
    stock_df['200_SMA'] = round(stock_df['Close'].rolling(window = 200, min_periods = 1).mean(),0)
    stock_df['Dev%_200'] = round((stock_df['Close'] - stock_df['200_SMA'])*100/stock_df['200_SMA'],2)
    stock_df.dropna(axis = 0, inplace = True) # remove any null rows

    stock_df['20_SMA'] = stock_df['Close'].rolling(window=20).mean()
    stock_df['50_SMA'] = stock_df['Close'].rolling(window=50).mean()
    stock_df['Symbol'] = stock_symbol

    stock_df['Close'] = round(stock_df['Close'],2)
    stock_df['Min'] = round(min(stock_df['Close']),2)
    stock_df['RSI_14'] = ta.momentum.RSIIndicator(close=stock_df['Close'], window=14).rsi()
    stock_df['RSI_14'] = round(stock_df['RSI_14'],0)
    stock_df['Prev_Close'] = stock_df['Close'].shift(1)
    stock_df.drop(['Open', 'Low', 'High', 'Volume'], axis=1, inplace=True)
    stock_df = stock_df.tail(1)

    max_SMA = max(stock_df['20_SMA'].item(), stock_df['50_SMA'].item(), stock_df['200_SMA'].item())
    min_SMA = min(stock_df['20_SMA'].item(), stock_df['50_SMA'].item(), stock_df['200_SMA'].item())
    ABS_Spread = max_SMA - min_SMA
    stock_df['Spread%'] = round((ABS_Spread / stock_df['200_SMA'].item()) * 100,2)

    return stock_df

# function to compute stock attributes
def get_common_features(stock_ids, df_mypf):

  df_prec_dev = pd.DataFrame()
  df_tmp = get_screener_features()
  df_rs = get_relative_strength(stock_ids)

  for stock_id in stock_ids:
      tmp = stock_prec_dev(stock_id)
      tmp = tmp.reset_index()
      df_prec_dev = pd.concat([df_prec_dev, tmp], ignore_index = True)
  df_prec_dev = pd.merge(df_prec_dev, df_mypf, on= 'Symbol')
  df_prec_dev = pd.merge(df_prec_dev, df_tmp, on= 'Symbol', how='left')
  df_prec_dev = pd.merge(df_prec_dev, df_rs, on= 'Symbol', how='left')
  #print(df_prec_dev.columns)
  df_prec_dev['Curr_PE'] = round(df_prec_dev['Close']/df_prec_dev['EPS'],1)
  df_prec_dev['Dev%_PE'] = round((df_prec_dev['Curr_PE'] - df_prec_dev['MedPE'])*100/df_prec_dev['MedPE'],2)
  df_prec_dev['Conviction'] = df_prec_dev['Conviction'] + '-' + df_prec_dev['CapType']
  return df_prec_dev

# function to arrange stock features
def arrange_features(df_stocks, common_cols, diff_cols):
  df_stocks_common = df_stocks[common_cols].drop_duplicates()
  df_stocks_diff = df_stocks[diff_cols]
  df_stocks_diff['Investment'] = df_stocks_diff['AvgCost'] * df_stocks_diff['Shares']
  df_stocks_diff = df_stocks_diff.groupby(['Symbol'])[['Shares', 'Investment']].aggregate(['sum']).reset_index()
  df_stocks_diff.columns = ['Symbol', 'Shares', 'Investment']
  df_stocks_diff['AvgCost'] = round(df_stocks_diff['Investment']/df_stocks_diff['Shares'],2)
  df_stocks = pd.merge(df_stocks_diff, df_stocks_common, on='Symbol')
  return df_stocks

# plot fact distribution across dimension
def plot_pie_chart(df, dimension, fact):
  # grouping the data by category and calculating the sum of fact for each type
  grouped_data = df.groupby(dimension)[fact].sum()

  # sorting the grouped data in descending order
  grouped_data = grouped_data.sort_values(ascending=False)

  # creating a pie chart
  grouped_data.plot.pie(autopct='%1.1f%%', startangle=90, figsize=(6, 6))

  # adding a title and displaying the plot
  plt.title(f'{dimension} {fact} Distribution')
  plt.ylabel('')  # To hide the y-label
  plt.show()
```

```python
# Portfolio feature engineering + snapshot metrics
def get_portfolio_features(df_common_features):
    print_date_time()
    print('=================== PORTFOLIO QUALITY ====================')

    df_common_features['NTT'] = np.where(
        df_common_features['Strategy'] == 'NTT',
        df_common_features["Target"],
        df_common_features["Max"]
    )
    df_common_features['LTT'] = np.where(
        df_common_features['Strategy'] == 'BTT',
        df_common_features["Target"],
        df_common_features["Max"]
    )
    df_common_features['BOL'] = df_common_features['Min']

    tmp_df = df_common_features[
        df_common_features['Symbol'].isin(
            mypf_df[mypf_df['InPortfolio'] != 'NA'].Symbol.values
        )
    ]
    print('qualified stocks: ' + str(len(tmp_df['Symbol'].unique())))

    tmp_df1 = tmp_df[tmp_df['LatestQtr'] == 1]
    print('with latest results: ' + str(len(tmp_df1['Symbol'].unique())))

    tmp_df1 = tmp_df1[tmp_df1['StarStock'] == 1]
    print('still star stocks: ' + str(len(tmp_df1['Symbol'].unique())))

    tmp_df['Investment'] = tmp_df['AvgCost'] * tmp_df['Shares']
    tmp_df['Current'] = round(tmp_df['Close'] * tmp_df['Shares'], 0)
    tmp_df['Previous'] = tmp_df['Prev_Close'] * tmp_df['Shares']
    tmp_df['EstimatedST'] = tmp_df['NTT'] * tmp_df['Shares']
    tmp_df['EstimatedLT'] = tmp_df['LTT'] * tmp_df['Shares']

    tmp_df['Current P/L'] = round(tmp_df['Current'] - tmp_df['Investment'], 0)
    tmp_df['Today P/L%'] = round(
        (tmp_df['Current'] - tmp_df['Previous']) * 100 / tmp_df['Previous'], 2
    )
    tmp_df['Current P/L%'] = round(
        (tmp_df['Current'] - tmp_df['Investment']) * 100 / tmp_df['Investment'], 2
    )
    tmp_df['EstimatedST P/L%'] = round(
        (tmp_df['EstimatedST'] - tmp_df['Investment']) * 100 / tmp_df['Investment'], 2
    )
    tmp_df['EstimatedLT P/L%'] = round(
        (tmp_df['EstimatedLT'] - tmp_df['Investment']) * 100 / tmp_df['Investment'], 2
    )
    tmp_df['NTT%'] = round(
        (tmp_df['NTT'] - tmp_df['Close']) * 100 / tmp_df['Close'], 2
    )
    tmp_df['LTT%'] = round(
        (tmp_df['LTT'] - tmp_df['Close']) * 100 / tmp_df['Close'], 2
    )
    tmp_df['Gained%'] = round(
        (tmp_df['Close'] - tmp_df['BOL']) * 100 / tmp_df['BOL'], 2
    )

    investment = round(sum(tmp_df['AvgCost'] * tmp_df['Shares']), 0)
    current = round(sum(tmp_df['Close'] * tmp_df['Shares']), 0)

    tmp_df['InitAlloc%'] = round(tmp_df['Investment'] * 100 / investment, 2)
    tmp_df['CurrAlloc%'] = round(tmp_df['Current'] * 100 / current, 2)

    tmp_df['FTT'] = tmp_df['LTT']
    tmp_df.loc[tmp_df['Strategy'] == 'NTT', 'FTT'] = tmp_df['NTT']
    tmp_df['FTT%'] = tmp_df['LTT%']
    tmp_df.loc[tmp_df['Strategy'] == 'NTT', 'FTT%'] = tmp_df['NTT%']
    tmp_df['FTT Amt'] = round(tmp_df['FTT%'] * tmp_df['Current'] / 100, 0)

    tmp_df['OTT%'] = round(
        (tmp_df['FTT'] - tmp_df['AvgCost']) * 100 / tmp_df['AvgCost'], 2
    )
    tmp_df['RRR Ind'] = round(tmp_df['Current P/L'] / tmp_df['FTT Amt'], 2)
    tmp_df['Risk Ind'] = round(tmp_df['Current P/L%'] * tmp_df['CurrAlloc%'], 0)

    return tmp_df


def build_portfolio_snapshot(df_portfolio_features, myinvmt_df):
    """Return the exact portfolio-level metrics used by the original print function."""
    from pyxirr import xirr

    total_booked_amt, reserve_amt, cy_booked_amt, py_booked_amt = get_amt()

    dates = myinvmt_df['Date'].values
    dates = np.append(dates, datetime.date.today().strftime('%d-%b-%y'))
    investment_history = myinvmt_df['Investment'].values
    dates = pd.to_datetime(dates)

    current = round(
        sum(df_portfolio_features['Close'] * df_portfolio_features['Shares']), 0
    ) + reserve_amt

    investment_xirr = np.append(investment_history, current)
    cagr = round(
        xirr(pd.DataFrame({"dates": dates, "amounts": investment_xirr})) * 100, 2
    )

    investment = -sum(investment_history)
    invested = round(
        sum(df_portfolio_features['AvgCost'] * df_portfolio_features['Shares']), 0
    ) + reserve_amt
    previous = round(
        sum(df_portfolio_features['Prev_Close'] * df_portfolio_features['Shares']), 0
    ) + reserve_amt
    cy_invested = investment + py_booked_amt

    today_pnl_amount = current - previous
    today_pnl_percentage = round((current - previous) * 100 / previous, 2)

    curr_pnl_amount = current - invested
    curr_pnl_percentage = round(curr_pnl_amount * 100 / cy_invested, 2)

    cy_pnl_amount = cy_booked_amt + curr_pnl_amount
    cy_pnl_percentage = round(cy_pnl_amount * 100 / cy_invested, 2)

    overall_pnl_amount = total_booked_amt + curr_pnl_amount
    overall_pnl_percentage = round(overall_pnl_amount * 100 / investment, 2)

    estimate_st = round(
        sum(df_portfolio_features['FTT'] * df_portfolio_features['Shares']), 0
    ) + reserve_amt
    est_st_pnl_amount = estimate_st - current
    est_st_pnl_percentage = round(est_st_pnl_amount * 100 / current, 2)

    estimate_lt = round(
        sum(df_portfolio_features['LTT'] * df_portfolio_features['Shares']), 0
    ) + reserve_amt
    est_lt_pnl_amount = estimate_lt - current
    est_lt_pnl_percentage = round(est_lt_pnl_amount * 100 / current, 2)

    total_profit = round(
        sum(df_portfolio_features[df_portfolio_features['Current P/L%'] > 0]['Current'])
        - sum(df_portfolio_features[df_portfolio_features['Current P/L%'] > 0]['Investment']),
        0
    )
    total_loss = round(
        sum(df_portfolio_features[df_portfolio_features['Current P/L%'] < 0]['Current'])
        - sum(df_portfolio_features[df_portfolio_features['Current P/L%'] < 0]['Investment']),
        0
    )

    return {
        "deployed_value": investment,
        "current_value": current,
        "xirr": cagr,
        "est_ftt_value": estimate_st,
        "future_upside_amt": est_st_pnl_amount,
        "future_upside_pct": est_st_pnl_percentage,
        "est_lt_value": estimate_lt,
        "est_lt_upside_amt": est_lt_pnl_amount,
        "est_lt_upside_pct": est_lt_pnl_percentage,
        "current_pnl_amt": curr_pnl_amount,
        "current_pnl_pct": curr_pnl_percentage,
        "total_booked_pnl_amt": total_booked_amt,
        "booked_plus_current_amt": overall_pnl_amount,
        "booked_plus_current_pct": overall_pnl_percentage,
        "cy_booked_pnl_amt": cy_booked_amt,
        "cy_booked_pct": round(cy_booked_amt * 100 / current, 2),
        "prev_year_booked_amt": py_booked_amt,
        "prev_year_booked_pct": round(py_booked_amt * 100 / investment, 2),
        "today_pnl_amt": today_pnl_amount,
        "today_pnl_pct": today_pnl_percentage,
        "cy_invested": cy_invested,
        "reserve": reserve_amt,
        "total_profit": total_profit,
        "total_loss": total_loss,
        "portfolio_count": df_portfolio_features['Symbol'].nunique(),
    }


def print_portfolio_stats(df_portfolio_features, myinvmt_df):
    """Backward-compatible text snapshot; also returns metrics for the UI layer."""
    m = build_portfolio_snapshot(df_portfolio_features, myinvmt_df)

    print('=================== PORTFOLIO SNAPSHOT ===================')
    print('Deployed Value: ', format_amt(m["deployed_value"]))
    print('Current Value:  ', format_amt(m["current_value"]))
    print('CAGR/XIRR %:    ', '{}%'.format(m["xirr"]))
    print('Est FTT Value:  ', format_amt(m["est_ftt_value"]))
    print('Future Upside:  ', '{} ({}%)'.format(
        format_amt(m["future_upside_amt"]), m["future_upside_pct"]
    ))

    print('=================== PERFORMANCE ==========================')
    print('Current PnL:             {} ({}%)'.format(
        format_amt(m["current_pnl_amt"]), m["current_pnl_pct"]
    ))
    print('Total Booked PnL:        {} ({}%)'.format(
        format_amt(m["total_booked_pnl_amt"]),
        round(m["total_booked_pnl_amt"] * 100 / m["deployed_value"], 2)
    ))
    print('Booked + Current PnL:    {} ({}%)'.format(
        format_amt(m["booked_plus_current_amt"]), m["booked_plus_current_pct"]
    ))
    print('CY Booked PnL:           {} ({}%)'.format(
        format_amt(m["cy_booked_pnl_amt"]), m["cy_booked_pct"]
    ))
    print('Prev Year Booked PnL:    {} ({}%)'.format(
        format_amt(m["prev_year_booked_amt"]), m["prev_year_booked_pct"]
    ))

    print('=================== TODAY / OPERATIONS ===================')
    print('Today PnL:               {} ({}%)'.format(
        format_amt(m["today_pnl_amt"]), m["today_pnl_pct"]
    ))
    print('CY Investment:           ', format_amt(m["cy_invested"]))
    print('Reserve:                 ', format_amt(m["reserve"]))

    print('=================== P&L DETAIL ============================')
    print('Total Profit:            ', format_amt(m["total_profit"]))
    print('Total Loss:              ', format_amt(m["total_loss"]))
    print('============================================================')

    return m
```

```python
# ============================================================
# 1. DATA LOAD + EXISTING QUANTVESTING CALCULATION ENGINE
# ============================================================
# Data continues to come directly from the existing GitHub repo.
# No portfolio source, calculation, threshold or strategy logic is changed.

start_date, end_date = get_start_end_date()

mypfs_df = get_mypfs_df()
mypps_df = get_mypps_df()
myinvmt_df = get_myinvmt_df()

mypf_df = pd.merge(mypfs_df, mypps_df, on="Symbol")

dm_pf = mypf_df[mypf_df['InPortfolio'] == 'DM']
sv_pf = mypf_df[mypf_df['InPortfolio'] == 'SV']

dm_stocks = get_stock_ids(dm_pf)
sv_stocks = get_stock_ids(sv_pf)
df_stocks = pd.concat([dm_pf, sv_pf], ignore_index=True)

common_cols = [
    'Symbol', 'Target', 'Criteria', 'Strategy', 'CumlRnk',
    'LatestQtr', 'StarStock', 'Conviction', 'Category',
    'InFolio', 'XIRR', 'MBQ'
]
diff_cols = ['Symbol', 'AvgCost', 'Shares']

df_stocks = arrange_features(df_stocks, common_cols, diff_cols)

stock_ids = df_stocks['Symbol'].values
df_common_features = get_common_features(stock_ids, df_stocks)

df_common_features.reset_index(drop=True, inplace=True)
df_common_features.drop(['Date'], axis=1, inplace=True)

df_portfolio_features = get_portfolio_features(df_common_features)
portfolio_snapshot = print_portfolio_stats(df_portfolio_features, myinvmt_df)

df = df_portfolio_features
list_ox40n = list(
    df[df['MBQ'].str.contains('OX40N', na=False)]['Symbol'].values
)

print("\nCalculation engine complete. Launching the intelligence layer below...")
```

```python
# ============================================================
# UI HELPERS — presentation only; no investment logic changed
# ============================================================

DISPLAY_COLS = [
    'Symbol', 'Current', 'CurrAlloc%', 'Current P/L', 'Current P/L%',
    'Today P/L%', 'FTT Amt', 'FTT%', 'RSI_14', 'Dev%_PE',
    'Spread%', 'CumlRnk', 'RRR Ind', 'Gained%', 'Conviction',
    'Criteria', 'Strategy', 'Category'
]

def _available_cols(df, cols=DISPLAY_COLS):
    return [c for c in cols if c in df.columns]

def _money(x):
    try:
        return format_amt(float(x))
    except Exception:
        return str(x)

def _pct(x):
    try:
        return f"{float(x):,.2f}%"
    except Exception:
        return str(x)

def _signed_pct(x):
    try:
        return f"{float(x):+,.2f}%"
    except Exception:
        return str(x)

def _card(title, value, subtitle="", accent=""):
    border = f"border-top:4px solid {accent};" if accent else "border-top:4px solid #5b6472;"
    return f"""
    <div style="
        flex:1; min-width:180px; padding:18px 20px; margin:6px;
        border:1px solid #e5e7eb; border-radius:14px; {border}
        background:#ffffff; box-shadow:0 2px 10px rgba(0,0,0,.05);
        font-family:Inter,Arial,sans-serif;">
        <div style="font-size:12px; color:#6b7280; text-transform:uppercase;
                    letter-spacing:.08em; font-weight:700;">{title}</div>
        <div style="font-size:26px; font-weight:800; margin:8px 0 4px;">{value}</div>
        <div style="font-size:12px; color:#6b7280;">{subtitle}</div>
    </div>
    """

def _section(title, subtitle=""):
    return HTML(f"""
    <div style="margin:24px 4px 10px; font-family:Inter,Arial,sans-serif;">
        <div style="font-size:22px;font-weight:800;color:#111827;">{title}</div>
        <div style="font-size:13px;color:#6b7280;margin-top:4px;">{subtitle}</div>
    </div>
    """)

def _indian_number(x):
    try:
        x = float(x)
        sign = '-' if x < 0 else ''
        x = abs(round(x))
        s = str(int(x))
        if len(s) <= 3:
            return sign + s
        last3 = s[-3:]
        remaining = s[:-3]
        parts = []
        while len(remaining) > 2:
            parts.insert(0, remaining[-2:])
            remaining = remaining[:-2]
        if remaining:
            parts.insert(0, remaining)
        return sign + ','.join(parts + [last3])
    except Exception:
        return str(x)

def _show_table(df, title=None, max_rows=10):
    if title:
        display(Markdown(f"#### {title}"))
    if df is None or df.empty:
        display(HTML('<div style="padding:14px;border:1px dashed #cbd5e1;border-radius:12px;color:#64748b;">No candidates under the current rule.</div>'))
        return

    view = df.head(max_rows).copy()

    # Table amounts: Indian comma grouping, no decimal places.
    amount_cols = [
        'Current', 'Current P/L', 'FTT Amt', 'Investment',
        'Previous', 'EstimatedST', 'EstimatedLT'
    ]

    for c in amount_cols:
        if c in view.columns:
            view[c] = view[c].map(_indian_number)

    # Percentages: exactly 2 decimal places.
    for c in [c for c in view.columns if '%' in c]:
        view[c] = view[c].map(_pct)

    # Integer metrics: no decimal places.
    for c in ['RSI_14', 'CumlRnk']:
        if c in view.columns:
            view[c] = view[c].map(
                lambda x: f"{float(x):.0f}" if pd.notna(x) else "—"
            )

    # RRR: exactly 2 decimal places.
    if 'RRR Ind' in view.columns:
        view['RRR Ind'] = view['RRR Ind'].map(
            lambda x: f"{float(x):.2f}" if pd.notna(x) else "—"
        )

    cols = _available_cols(view)
    if not cols:
        cols = list(view.columns)
    display(view[cols].style.hide(axis="index"))

def _plotly_layout(fig, title):
    fig.update_layout(
        title=title,
        template="plotly_white",
        margin=dict(l=20,r=20,t=55,b=20),
        height=390,
        font=dict(family="Inter, Arial, sans-serif"),
    )
    return fig
```

```python
# ============================================================
# ACTION LIBRARY — preserves the original notebook's insight rules
# ============================================================

def build_action_views(df):
    views = {}

    # Original insight #1 — top 5 near their targets
    tmp = df[~df['Symbol'].isin(['ENRIN','BLUSPRING','DIGITIDE'])].sort_values(
        by='FTT Amt', ascending=True
    )
    views["Near Target"] = tmp.head(5).copy()

    # Original insight #2 — accumulate
    tmp = df[df['Conviction'].isin(['X-LC','H-LC'])]
    tmp = tmp[(tmp['CumlRnk'] < 40) & (tmp['FTT%'] > 20)].sort_values(
        by='CurrAlloc%', ascending=True
    )
    views["Accumulate"] = tmp.head(5).copy()

    # Original insight #3 — today's winners
    views["Today's Winners"] = df.sort_values(by='Today P/L%', ascending=False).head(5).copy()

    # Original insight #4 — today's laggards
    views["Today's Laggards"] = df.sort_values(by='Today P/L%', ascending=True).head(5).copy()

    # Original insight #5 — exit based on cumulative rank
    tmp = df[~df['Conviction'].isin(['X-LC','H-LC','X-MC','X-SC'])]
    tmp = tmp[(tmp['Current P/L%'] > -1) & (tmp['Current P/L%'] < 1)].sort_values(
        by='CumlRnk', ascending=False
    )
    views["Exit Review — Rank"] = tmp.head(5).copy()

    # Original insight #6 — OX40N
    views["OX40N Exit Review"] = df[df['Symbol'].isin(list_ox40n)].sort_values(
        by='Current P/L%', ascending=False
    ).head(5).copy()

    # Original insight #7 — book for rotation
    views["Book for Rotation"] = df[df['Current P/L%'] > 20].sort_values(
        by='CumlRnk', ascending=False
    ).head(5).copy()

    # Original insight #8 — monitor for exit
    tmp = df[~df['Conviction'].isin(['X-LC','H-LC','X-MC','X-SC'])]
    tmp = tmp[tmp['Current P/L%'] > 0].sort_values(by='Dev%_PE', ascending=False)
    views["Monitor Exit"] = tmp.head(5).copy()

    # Original insight #9 — weak conviction rotation
    tmp = df[~df['Conviction'].isin(['X-LC','H-LC','X-MC','X-SC'])]
    tmp = tmp[~tmp['Criteria'].isin(['XY25','XY24','X40','X40N'])].sort_values(
        by='RRR Ind', ascending=False
    )
    views["Weak Conviction Rotation"] = tmp.head(5).copy()

    # Original insight #10 — lowest RSI
    tmp = df[df['Conviction'].isin(['X-LC','H-LC','X-MC','X-SC'])].sort_values(
        by='RSI_14', ascending=True
    )
    views["Accumulate — RSI"] = tmp.head(5).copy()

    # Original insight #11 — spread + rank
    tmp = df[df['Conviction'].isin(['X-LC','H-LC','X-MC','X-SC'])]
    tmp = tmp[tmp['CumlRnk'] < 100].sort_values(by='Spread%', ascending=True)
    views["Accumulate — Spread"] = tmp.head(5).copy()

    # Original insight #12 — lowest gained
    views["Accumulate — Gained"] = df[
        df['Conviction'].isin(['X-LC','H-LC','X-MC','X-SC'])
    ].sort_values(by='Gained%', ascending=True).head(5).copy()

    # Original insight #13 — lowest allocation
    views["Accumulate — Allocation"] = df[
        df['Conviction'].isin(['X-LC','H-LC','X-MC','X-SC'])
    ].sort_values(by='CurrAlloc%', ascending=True).head(5).copy()

    # Original insight #14 — cumulative rank
    views["Best Cumulative Rank"] = df.sort_values(by='CumlRnk', ascending=True).head(5).copy()

    # Original insight #15 — average up
    views["Average Up"] = df[df['Dev%_200'] > 5].sort_values(
        by='CurrAlloc%', ascending=True
    ).head(5).copy()

    # Original insight #16 — relative strength percentile
    views["Relative Strength"] = df.sort_values(by='RSP', ascending=False).head(5).copy()

    return views

ACTION_VIEWS = build_action_views(df_portfolio_features)

ACTION_META = {
    "Near Target": ("Review target proximity", "Prioritise holdings where the existing FTT amount is smallest."),
    "Accumulate": ("Review accumulation candidates", "Existing high-conviction + rank + upside filters are satisfied."),
    "Today's Winners": ("Review today's strength", "Largest positive daily P/L within the current portfolio."),
    "Today's Laggards": ("Review today's weakness", "Largest negative daily P/L within the current portfolio."),
    "Exit Review — Rank": ("Review exit candidates", "Weak conviction + near-flat P/L + weaker cumulative rank."),
    "OX40N Exit Review": ("Review OX40N names", "Existing OX40N portfolio rule is triggered."),
    "Book for Rotation": ("Review profit rotation", "Existing rule flags holdings with >20% current P/L."),
    "Monitor Exit": ("Monitor for exit", "Positive P/L but weaker conviction and elevated PE deviation."),
    "Weak Conviction Rotation": ("Review rotation", "Weak conviction outside the excluded criteria, ranked by RRR."),
    "Accumulate — RSI": ("Review RSI pullbacks", "High-conviction names with the lowest RSI."),
    "Accumulate — Spread": ("Review technical spread", "High-conviction names with lower SMA spread and rank < 100."),
    "Accumulate — Gained": ("Review lower-gained names", "High-conviction names with lower Gained%."),
    "Accumulate — Allocation": ("Review under-allocation", "High-conviction names with the lowest current allocation."),
    "Best Cumulative Rank": ("Review strongest rank", "Lowest cumulative rank in the existing framework."),
    "Average Up": ("Review average-up candidates", "Existing Dev%_200 > 5 filter, prioritised by lower allocation."),
    "Relative Strength": ("Review relative strength", "Highest RSP percentile within the portfolio."),
}

ACTION_VIEW_OPTIONS = list(ACTION_VIEWS.keys())
```

## 4. Stock Drill-down

Select any holding to move from the portfolio-level signal to the evidence behind that stock.


```python
# ============================================================
# STOCK DRILL-DOWN
# ============================================================

stock_selector = widgets.Dropdown(
    options=sorted(df_portfolio_features['Symbol'].dropna().unique().tolist()),
    description="Holding:",
    layout=widgets.Layout(width="420px"),
    style={"description_width": "75px"}
)
stock_output = widgets.Output()

def render_stock(change=None):
    with stock_output:
        clear_output(wait=True)
        symbol = stock_selector.value
        row = df_portfolio_features[df_portfolio_features['Symbol'] == symbol].iloc[0]

        cards = [
            _card("Current", _money(row.get('Current', np.nan)), f"Allocation {_pct(row.get('CurrAlloc%', np.nan))}"),
            _card("Current P/L", _pct(row.get('Current P/L%', np.nan)), _money(row.get('Current P/L', np.nan))),
            _card("FTT", _pct(row.get('FTT%', np.nan)), _money(row.get('FTT Amt', np.nan))),
            _card("RSI", str(row.get('RSI_14', '—')), "14-period"),
            _card("Rank", str(row.get('CumlRnk', '—')), "Cumulative rank"),
        ]
        display(HTML('<div style="display:flex;flex-wrap:wrap;margin:-6px;">' + ''.join(cards) + '</div>'))

        evidence_cols = [
            'Symbol','Conviction','Criteria','Strategy','Category','AvgCost','Shares',
            'Close','Target','NTT','LTT','CurrAlloc%','Current P/L%','Today P/L%',
            'FTT%','Dev%_PE','Dev%_200','Spread%','Gained%','RSP','RRR Ind','Risk Ind'
        ]
        evidence_cols = [c for c in evidence_cols if c in row.index]
        evidence = pd.DataFrame([row[evidence_cols].to_dict()])
        evidence = evidence.T.rename(columns={0:'Value'})

        if 'Value' in evidence.columns:
            if 'RSI_14' in evidence.index:
                evidence.loc['RSI_14', 'Value'] = f"{float(evidence.loc['RSI_14', 'Value']):.0f}"
            if 'CumlRnk' in evidence.index:
                evidence.loc['CumlRnk', 'Value'] = f"{float(evidence.loc['CumlRnk', 'Value']):.0f}"
            if 'RRR Ind' in evidence.index:
                evidence.loc['RRR Ind', 'Value'] = f"{float(evidence.loc['RRR Ind', 'Value']):.2f}"

            for metric in evidence.index:
                if '%' in metric:
                    try:
                        evidence.loc[metric, 'Value'] = f"{float(evidence.loc[metric, 'Value']):.2f}%"
                    except Exception:
                        pass

            for metric in ['AvgCost', 'Close', 'Target', 'NTT', 'LTT']:
                if metric in evidence.index:
                    evidence.loc[metric, 'Value'] = _indian_number(evidence.loc[metric, 'Value'])

        display(evidence)

stock_selector.observe(render_stock, names="value")
display(stock_selector)
display(stock_output)
render_stock()
```
