# Quantvesting | Start Here

## A customer-first journey from portfolio data to decision intelligence

Quantvesting is a notebook-first investment-analysis and decision-support framework built around a curated universe of quality Indian businesses, valuation discipline, FTT-based opportunity assessment and portfolio intelligence.

**The goal is not to predict markets. The goal is to help an investor understand their portfolio, identify what deserves attention, understand remaining opportunity and make better-informed decisions.**

## Recommended customer journey

```text
00 Start Here
      ↓
01 Onboard My Portfolio
      ↓
02 My Portfolio Health
      ↓
03 My Portfolio Decisions
      ↓
04 Quantvesting Prospects
      ↓
05 Capital Rotation
      ↓
06 My Quantvesting Journey
```

## What a new customer needs

The first assessment is designed around a minimum portfolio file:

```csv
Symbol,Shares,AvgCost
TCS,100,3200
INFY,150,1450
HDFCBANK,200,1650
```

Account labels and investment history are optional. The onboarding notebook normalizes common broker/export column names into the existing Quantvesting portfolio contract.

## What Quantvesting can help answer

- **Portfolio Health:** What do I own and how is it distributed?
- **Portfolio Decisions:** What deserves attention right now?
- **Prospects:** What does the Quantvesting universe currently look like?
- **Capital Rotation:** Where has the thesis been substantially captured relative to available opportunity?
- **Journey:** How has my portfolio evolved over time?

## Important architecture principle

```text
Customer input / market data
          ↓
   Quantvesting Engine
          ↓
 Structured analytical output
          ↓
 Jupyter presentation today
 API / Web later
```

**Notebooks present the analysis; the engine performs the calculations.**

## Existing methodology remains intact

The current journey exposes the established Quantvesting framework including X/H/M/L quality, LC/MC/SC categorisation, the full conviction priority map, CumlRnk, FTT / NTT / LTT / BOL, RRR Ind, Risk Ind, portfolio analytics, prospect analysis, capital-rotation review, Phase-B run tracking, validation and reproducibility.
